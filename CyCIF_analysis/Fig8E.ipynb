```python
from scipy.stats import chi2_contingency

# Rows: entropy (High, Low) | Columns: proportion of IRF1+ tumor cells (0, Non-0)
data_stats = np.array([[6, 14],   # High entropy
                 [15,  5]]) # Low entropy

chi2, p, dof, expected = chi2_contingency(data_stats)

print(f"Chi-Square Statistic: {chi2:.4f}")
print(f"P-value: {p:.4f}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:\n", expected)
```
